# Process Weekly Data - Simple Unified Approach

Drop all your weekly exports in a folder. This agent figures it out.

````python
import os
from pathlib import Path
from anthropic import Anthropic
from dotenv import load_dotenv

load_dotenv()
client = Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))
````

## One Function - Does Everything

Point it at a folder with:
- Chat exports (any format)
- Medical records (any format)
- Work data (any format)
- Research notes
- Anything else

Gets you:
- Timeline in Obsidian
- All events extracted
- Everything cross-referenced

````python
async def process_weekly_folder(folder_path: str, week_name: str = None):
    """
    Process entire weekly data folder.
    
    Args:
        folder_path: Path to folder with all your weekly data
        week_name: Optional name (auto-detected from folder)
    
    Returns:
        Timeline events and Obsidian export
    """
    
    folder = Path(folder_path)
    week_name = week_name or folder.name
    
    print(f"Processing {week_name}...\n")
    
    # Get all files
    all_files = list(folder.rglob("*"))
    files = [f for f in all_files if f.is_file() and not f.name.startswith(".")]
    
    print(f"Found {len(files)} files")
    
    # Read all content
    file_contents = {}
    for file in files:
        try:
            # Try to read as text
            content = file.read_text(encoding="utf-8", errors="ignore")
            if len(content) > 0:
                file_contents[file.name] = content[:50000]  # First 50KB
        except:
            # Binary file - skip for now
            file_contents[file.name] = f"[Binary file: {file.suffix}]"
    
    # Ask Claude to process everything at once
    prompt = f"""
You are processing weekly data exports for {week_name}.

I'm giving you {len(files)} files containing:
- Chat exports (ChatGPT, Claude, Perplexity)
- Medical records (appointments, labs, imaging, vitals)
- Work data (timesheets, incidents, communications)
- Research notes
- Any other relevant data

Your task:
1. Extract ALL events, incidents, appointments, and significant occurrences
2. For each event, identify:
   - What happened
   - When it happened
   - Source (which file/conversation)
   - Type (medical, work, personal, etc.)
   - Entities involved (people, places, organizations)

3. Build a chronological timeline
4. Identify cross-references (same event mentioned in multiple sources)
5. Output as structured JSON

Files:

"""
    
    for filename, content in file_contents.items():
        prompt += f"\n--- {filename} ---\n{content}\n"
    
    prompt += """

Return JSON:
{{
  "events": [
    {{
      "date": "2026-01-06",
      "time": "14:30",  // optional
      "description": "Severe back pain after lifting at work",
      "type": "medical_symptom",
      "source": "claude_conversations.json",
      "entities": ["Patient", "Workplace"],
      "confidence": 0.9,
      "related_events": [1, 3]  // indices of related events
    }}
  ],
  "summary": "Brief summary of the week"
}}
"""
    
    print("\nAsking Claude to process all files...")
    
    response = client.messages.create(
        model="claude-sonnet-4-5-20250929",
        max_tokens=16000,
        messages=[{"role": "user", "content": prompt}]
    )
    
    # Parse response
    import json
    result_text = response.content[0].text
    
    # Extract JSON from markdown if present
    if "```json" in result_text:
        result_text = result_text.split("```json")[1].split("```")[0].strip()
    elif "```" in result_text:
        result_text = result_text.split("```")[1].split("```")[0].strip()
    
    timeline = json.loads(result_text)
    
    print(f"\n✓ Extracted {len(timeline['events'])} events\n")
    
    # Export to Obsidian
    export_to_obsidian(timeline, week_name)
    
    return timeline


def export_to_obsidian(timeline: dict, week_name: str):
    """Create Obsidian notes from timeline."""
    
    vault_dir = Path("obsidian_vault") / week_name
    vault_dir.mkdir(parents=True, exist_ok=True)
    
    # Create timeline note
    md = f"""---
type: timeline
week: {week_name}
events: {len(timeline['events'])}
---

# {week_name} Timeline

{timeline.get('summary', '')}

## Events

"""
    
    for i, event in enumerate(timeline['events']):
        time_str = event.get('time', '')
        md += f"""### {event['date']} {time_str} - {event['description']}
- **Type**: {event['type']}
- **Source**: {event['source']}
- **Entities**: {', '.join(event.get('entities', []))}
- **Confidence**: {event.get('confidence', 0):.2f}

"""
        
        # Cross-references
        related = event.get('related_events', [])
        if related:
            md += "**Related events**:\n"
            for rel_idx in related:
                if rel_idx < len(timeline['events']):
                    rel = timeline['events'][rel_idx]
                    md += f"- {rel['date']}: {rel['description']}\n"
            md += "\n"
    
    # Write timeline
    (vault_dir / "Timeline.md").write_text(md)
    
    print(f"✓ Exported to {vault_dir}/Timeline.md")
````

## Usage

Just point it at your folder:

````python
# Process one week
timeline = await process_weekly_folder("data/2026_week_01")

# View results
print(f"\nWeek Summary: {timeline['summary']}")
print(f"\nEvents:")
for event in timeline['events'][:5]:  # First 5
    print(f"  {event['date']}: {event['description']}")
````
